```python

# ================================================================
# MINNESOTA TIMBERWOLVES, 2025–26
# Edwards dependence and playmaking
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display
from google.colab import drive

drive.mount("/content/drive")

ROOT = Path("/content/drive/MyDrive/minnesota_analysis")
OUT = ROOT / "data/processed/category_2_final"
FIG = ROOT / "figures/category_2_final"

OUT.mkdir(parents=True, exist_ok=True)
FIG.mkdir(parents=True, exist_ok=True)

PHASES = ["Regular Season", "Playoffs"]
CUP_FINAL = "401809839"

plt.rcParams.update({
    "figure.figsize": (10, 5),
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False
})


# ================================================================
# 1. LOAD EXISTING ESPN FILES
# ================================================================

def locate_file(filename):
    matches = list(ROOT.rglob(filename))

    # Prefer the original raw-data copy.
    matches.sort(
        key=lambda x: (
            "data/raw" not in str(x),
            len(str(x))
        )
    )

    if not matches:
        raise FileNotFoundError(
            f"Cannot find {filename} under {ROOT}.\n"
            "Check the location of your existing parquet files."
        )

    print(f"Using {filename}: {matches[0]}")
    return matches[0]


players = pd.read_parquet(
    locate_file("player_box_2026.parquet")
)

teams = pd.read_parquet(
    locate_file("team_box_2026.parquet")
)


# ================================================================
# 2. HELPER FUNCTIONS
# ================================================================

def find_column(df, candidates, required=True):
    lookup = {str(c).lower(): c for c in df.columns}

    for name in candidates:
        if name.lower() in lookup:
            return lookup[name.lower()]

    if required:
        raise KeyError(
            f"None of these columns found: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


def numeric_column(df, candidates):
    col = find_column(df, candidates, required=False)

    if col is None:
        return pd.Series(np.nan, index=df.index)

    return pd.to_numeric(df[col], errors="coerce")


def parse_minutes(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip()

    if ":" in text:
        parts = text.split(":")

        try:
            return (
                float(parts[0])
                + float(parts[1]) / 60
            )
        except ValueError:
            return np.nan

    return pd.to_numeric(text, errors="coerce")


def safe_div(numerator, denominator):
    return numerator / denominator.replace(0, np.nan)


def save_table(df, filename):
    path = OUT / filename
    df.to_csv(path, index=False)

    print(f"\nSaved: {path.name}")
    display(df.round(2))

    return df


def bar_chart(
    df,
    metric,
    title,
    filename,
    label="player",
    highlight=None
):
    data = df.dropna(subset=[metric]).copy()

    if data.empty:
        print(f"Skipped chart — data unavailable: {title}")
        return

    data = data.sort_values(metric)

    colors = [
        "#f97316" if name == highlight else "#2563eb"
        for name in data[label]
    ]

    fig, ax = plt.subplots(
        figsize=(9, max(3.5, len(data) * 0.65))
    )

    bars = ax.barh(
        data[label],
        data[metric],
        color=colors
    )

    values = data[metric].to_numpy(dtype=float)
    spread = max(values.max() - min(0, values.min()), 1)

    for bar, value in zip(bars, values):
        ax.text(
            value + spread * 0.015,
            bar.get_y() + bar.get_height() / 2,
            f"{value:.2f}",
            va="center",
            fontsize=9
        )

    lower = min(0, values.min() - spread * 0.1)
    upper = values.max() + spread * 0.25

    ax.set_xlim(lower, upper)
    ax.set_xlabel(metric)
    ax.set_title(title, fontweight="bold")
    ax.grid(axis="x", alpha=0.15)

    fig.tight_layout()
    fig.savefig(
        FIG / filename,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)


# ================================================================
# 3. SEPARATE TEAM AND PLAYER STATISTIC MAPPINGS
# ================================================================



TEAM_STATS = {
    "PTS": ["team_score", "points", "pts"],
    "FGM": ["field_goals_made", "fgm"],
    "FGA": ["field_goals_attempted", "fga"],
    "3PM": [
        "three_point_field_goals_made",
        "three_pointers_made",
        "fg3m"
    ],
    "3PA": [
        "three_point_field_goals_attempted",
        "three_pointers_attempted",
        "fg3a"
    ],
    "FTA": ["free_throws_attempted", "fta"],
    "ORB": ["offensive_rebounds", "oreb"],
    "AST": ["assists", "ast"],
    "TOV": [
        "total_turnovers",
        "turnovers",
        "tov"
    ]
}

PLAYER_STATS = {
    "PTS": ["points", "pts"],
    "FGM": ["field_goals_made", "fgm"],
    "FGA": ["field_goals_attempted", "fga"],
    "3PM": [
        "three_point_field_goals_made",
        "three_pointers_made",
        "fg3m"
    ],
    "3PA": [
        "three_point_field_goals_attempted",
        "three_pointers_attempted",
        "fg3a"
    ],
    "FTA": ["free_throws_attempted", "fta"],
    "ORB": ["offensive_rebounds", "oreb"],
    "AST": ["assists", "ast"],
    "TOV": ["turnovers", "tov"]
}



# ================================================================
# 4. CLEAN TEAM DATA
# ================================================================

t = teams.copy()

t["game_id"] = t["game_id"].astype(str)
t["season_type"] = pd.to_numeric(
    t["season_type"], errors="coerce"
)

t = t[
    t["season_type"].isin([2, 3])
    & (t["game_id"] != CUP_FINAL)
].copy()

t["phase"] = t["season_type"].map({
    2: "Regular Season",
    3: "Playoffs"
})

t["team"] = (
    t["team_abbreviation"]
    .astype(str)
    .str.upper()
    .str.strip()
)

# Load ordinary team statistics.
for stat, aliases in TEAM_STATS.items():
    if stat != "TOV":
        t[stat] = numeric_column(t, aliases)


t["TOV"] = pd.to_numeric(
    t["total_turnovers"],
    errors="coerce"
).combine_first(
    pd.to_numeric(
        t["turnovers"],
        errors="coerce"
    )
)


min_ids = set(
    t.loc[t["team"] == "MIN", "game_id"]
)

relevant = t["game_id"].isin(min_ids)

required_team_stats = [
    "PTS", "FGM", "FGA", "3PM",
    "FTA", "ORB", "AST", "TOV"
]

missing = t.loc[
    relevant
    & t[required_team_stats].isna().any(axis=1),
    ["game_id", "team", "phase"]
    + required_team_stats
]

if not missing.empty:
    print("Games with incomplete team statistics:")
    display(missing)

    raise ValueError(
        "Some Minnesota games or opponent games "
        "still have missing statistics. "
        "Inspect the table above."
    )

if t.duplicated(["game_id", "team"]).any():
    raise ValueError(
        "Duplicate team-game records detected."
    )

game_phase = t[
    ["game_id", "phase"]
].drop_duplicates()

if game_phase["game_id"].duplicated().any():
    raise ValueError(
        "A game appears in multiple season phases."
    )

print("Team statistics validated successfully.")

# ================================================================
# 5. CLEAN PLAYER DATA
# ================================================================

p = players.copy()
p["game_id"] = p["game_id"].astype(str)

player_col = find_column(
    p,
    [
        "athlete_display_name",
        "athlete_name",
        "player_name",
        "display_name"
    ]
)

player_team_col = find_column(
    p,
    ["team_abbreviation", "team_abbrev"]
)

minutes_col = find_column(
    p,
    ["minutes", "minutes_played", "min"]
)

p["player"] = (
    p[player_col]
    .astype(str)
    .str.strip()
)

p["team"] = (
    p[player_team_col]
    .astype(str)
    .str.upper()
    .str.strip()
)

p["minutes"] = p[minutes_col].apply(parse_minutes)

for stat, aliases in PLAYER_STATS.items():
    p[stat] = numeric_column(p, aliases)

position_col = find_column(
    p,
    [
        "athlete_position_abbreviation",
        "position_abbreviation",
        "position"
    ],
    required=False
)

if position_col is not None:
    p["position"] = (
        p[position_col]
        .astype(str)
        .str.upper()
        .str.strip()
    )
else:
    p["position"] = np.nan

# Remove any pre-existing phase column before merging.
p = p.drop(columns=["phase"], errors="ignore")

p = p.merge(
    game_phase,
    on="game_id",
    how="inner",
    validate="many_to_one"
)

# Exclude DNPs.
p = p[p["minutes"] > 0].copy()

required_player_stats = [
    "PTS", "FGM", "FGA", "3PM",
    "3PA", "FTA", "AST", "TOV"
]

for stat in required_player_stats:
    if p[stat].isna().any():
        raise ValueError(
            f"Missing player {stat} values.\n"
            f"Candidate columns: {PLAYER_STATS[stat]}"
        )

if p.duplicated(
    ["game_id", "team", "player"]
).any():
    raise ValueError(
        "Duplicate player-game records detected."
    )


# ================================================================
# 6. BUILD MINNESOTA GAME DATASET
# ================================================================

min_games = t[t["team"] == "MIN"].copy()

opponent_cols = [
    "PTS", "FGA", "FTA", "ORB", "TOV"
]

opponents = t[
    t["game_id"].isin(min_games["game_id"])
    & (t["team"] != "MIN")
][
    ["game_id"] + opponent_cols
].copy()

opponents = opponents.rename(
    columns={
        col: f"OPP_{col}"
        for col in opponent_cols
    }
)

min_games = min_games.merge(
    opponents,
    on="game_id",
    how="left",
    validate="one_to_one"
)

if min_games[
    [f"OPP_{c}" for c in opponent_cols]
].isna().any().any():
    raise ValueError(
        "Some Minnesota games are missing opponent box scores."
    )

print("\nMINNESOTA GAME COUNTS")
display(
    min_games.groupby("phase")["game_id"]
    .nunique()
    .reindex(PHASES)
)

assert (
    min_games[
        min_games["phase"] == "Regular Season"
    ]["game_id"].nunique() == 82
), "Expected 82 regular-season games."

assert (
    min_games[
        min_games["phase"] == "Playoffs"
    ]["game_id"].nunique() == 12
), "Expected 12 playoff games."


# ================================================================
# 7. MINNESOTA WITH/WITHOUT EDWARDS
# ================================================================

ed_games = set(
    p[
        (p["team"] == "MIN")
        & (p["player"] == "Anthony Edwards")
    ]["game_id"]
)

min_games["availability"] = np.where(
    min_games["game_id"].isin(ed_games),
    "With Edwards",
    "Without Edwards"
)

# Estimated team and opponent possessions.
#
# These are box-score estimates, NOT official NBA
# possession-level ratings or true on/off-court splits.

min_games["MIN_POSS"] = (
    min_games["FGA"]
    + 0.44 * min_games["FTA"]
    - min_games["ORB"]
    + min_games["TOV"]
)

min_games["OPP_POSS"] = (
    min_games["OPP_FGA"]
    + 0.44 * min_games["OPP_FTA"]
    - min_games["OPP_ORB"]
    + min_games["OPP_TOV"]
)

min_games["POSS"] = (
    min_games["MIN_POSS"]
    + min_games["OPP_POSS"]
) / 2

availability = (
    min_games
    .groupby(
        ["phase", "availability"],
        as_index=False
    )
    .agg(
        GP=("game_id", "nunique"),
        PTS=("PTS", "sum"),
        OPP_PTS=("OPP_PTS", "sum"),
        FGM=("FGM", "sum"),
        FGA=("FGA", "sum"),
        TPM=("3PM", "sum"),
        AST=("AST", "sum"),
        POSS=("POSS", "sum")
    )
)

availability["ORTG"] = (
    100
    * availability["PTS"]
    / availability["POSS"]
)

availability["NETRTG"] = (
    100
    * (
        availability["PTS"]
        - availability["OPP_PTS"]
    )
    / availability["POSS"]
)

availability["eFG%"] = (
    100
    * (
        availability["FGM"]
        + 0.5 * availability["TPM"]
    )
    / availability["FGA"]
)

availability["APG"] = (
    availability["AST"]
    / availability["GP"]
)

availability = availability[
    [
        "phase",
        "availability",
        "GP",
        "ORTG",
        "NETRTG",
        "eFG%",
        "APG"
    ]
]

print("\n1. MINNESOTA WITH/WITHOUT EDWARDS")

save_table(
    availability,
    "01_minnesota_with_without_edwards.csv"
)

for phase in PHASES:
    subset = availability[
        availability["phase"] == phase
    ]

    for metric in ["ORTG", "NETRTG", "eFG%", "APG"]:
        bar_chart(
            subset,
            metric,
            f"Minnesota {metric}: {phase}",
            f"01_{phase}_{metric}.png",
            label="availability",
            highlight="With Edwards"
        )


# ================================================================
# 8. MINNESOTA PASSING — PREVIOUSLY TRANSCRIBED FIGURES
# ================================================================



passing_rows = [
    # phase, player, GP, MPG, actual APG, potential APG

    ("Regular Season", "Julius Randle",
     79, 33.0, 5.0, 8.7),

    ("Regular Season", "Anthony Edwards",
     61, 35.0, 3.7, 6.6),

    ("Regular Season", "Donte DiVincenzo",
     82, 30.4, 3.8, 6.0),

    ("Regular Season", "Ayo Dosunmu",
     24, 28.9, 3.5, 5.2),

    ("Regular Season", "Mike Conley",
     54, 18.4, 2.9, 5.1),

    ("Playoffs", "Julius Randle",
     12, 33.3, 2.8, 5.8),

    ("Playoffs", "Anthony Edwards",
     10, 32.4, 2.7, 5.7),

    ("Playoffs", "Ayo Dosunmu",
     10, 29.2, 4.1, 4.8),

    ("Playoffs", "Jaden McDaniels",
     12, 33.8, 2.4, 3.6),

    ("Playoffs", "Mike Conley",
     12, 14.0, 2.7, 3.5)
]

min_passing = pd.DataFrame(
    passing_rows,
    columns=[
        "phase",
        "player",
        "NBA_GP",
        "NBA_MPG",
        "NBA_APG",
        "POT_AST_PG"
    ]
)

save_table(
    min_passing,
    "minnesota_passing_from_screenshots.csv"
)


# ================================================================
# 9. EDWARDS: REGULAR SEASON VS PLAYOFFS
# ================================================================

ed = p[
    (p["team"] == "MIN")
    & (p["player"] == "Anthony Edwards")
].copy()

ed_summary = (
    ed.groupby("phase", as_index=False)
    .agg(
        GP=("game_id", "nunique"),
        MIN=("minutes", "sum"),
        PTS=("PTS", "sum"),
        FGM=("FGM", "sum"),
        FGA=("FGA", "sum"),
        TPM=("3PM", "sum"),
        TPA=("3PA", "sum"),
        FTA=("FTA", "sum"),
        AST=("AST", "sum"),
        TOV=("TOV", "sum")
    )
)

ed_summary["MPG"] = (
    ed_summary["MIN"] / ed_summary["GP"]
)

ed_summary["PPG"] = (
    ed_summary["PTS"] / ed_summary["GP"]
)

ed_summary["FGA_PG"] = (
    ed_summary["FGA"] / ed_summary["GP"]
)

ed_summary["APG"] = (
    ed_summary["AST"] / ed_summary["GP"]
)

ed_summary["eFG%"] = (
    100
    * (
        ed_summary["FGM"]
        + 0.5 * ed_summary["TPM"]
    )
    / ed_summary["FGA"]
)

ed_summary["TS%"] = (
    100
    * ed_summary["PTS"]
    / (
        2
        * (
            ed_summary["FGA"]
            + 0.44 * ed_summary["FTA"]
        )
    )
)

ed_summary["3P%"] = (
    100
    * ed_summary["TPM"]
    / ed_summary["TPA"]
)

# Use Minnesota's totals ONLY in games Edwards played.
team_with_ed = (
    min_games[
        min_games["game_id"].isin(ed_games)
    ]
    .groupby("phase", as_index=False)
    .agg(
        TEAM_PTS=("PTS", "sum"),
        TEAM_FGA=("FGA", "sum")
    )
)

ed_summary = ed_summary.merge(
    team_with_ed,
    on="phase",
    how="left"
)

ed_summary["SCORING_SHARE%"] = (
    100
    * ed_summary["PTS"]
    / ed_summary["TEAM_PTS"]
)

ed_summary["SHOT_SHARE%"] = (
    100
    * ed_summary["FGA"]
    / ed_summary["TEAM_FGA"]
)

ed_passing = min_passing[
    min_passing["player"] == "Anthony Edwards"
][
    ["phase", "POT_AST_PG"]
]

ed_summary = ed_summary.merge(
    ed_passing,
    on="phase",
    how="left"
)

ed_output = ed_summary[
    [
        "phase",
        "GP",
        "MPG",
        "PPG",
        "FGA_PG",
        "POT_AST_PG",
        "APG",
        "eFG%",
        "TS%",
        "3P%",
        "SCORING_SHARE%",
        "SHOT_SHARE%"
    ]
]

print("\n2. EDWARDS: REGULAR SEASON VS PLAYOFFS")

save_table(
    ed_output,
    "02_edwards_regular_vs_playoffs.csv"
)

for metric in [
    "PPG",
    "FGA_PG",
    "POT_AST_PG",
    "APG",
    "eFG%",
    "TS%",
    "3P%",
    "SCORING_SHARE%",
    "SHOT_SHARE%"
]:
    bar_chart(
        ed_output,
        metric,
        f"Edwards: {metric}",
        f"02_edwards_{metric}.png",
        label="phase"
    )


# ================================================================
# 10. MINNESOTA TOP FIVE POTENTIAL-ASSIST CREATORS
# ================================================================

for phase, minimum_games in [
    ("Regular Season", 20),
    ("Playoffs", 5)
]:

    top = (
        min_passing[
            (min_passing["phase"] == phase)
            & (
                min_passing["NBA_GP"]
                >= minimum_games
            )
        ]
        .nlargest(5, "POT_AST_PG")
        .copy()
    )

    print(
        f"\n3. MINNESOTA TOP FIVE CREATORS: {phase}"
    )

    save_table(
        top,
        f"03_minnesota_top5_{phase}.csv"
    )

    x = np.arange(len(top))
    width = 0.36

    fig, ax = plt.subplots(figsize=(10, 5))

    ax.bar(
        x - width / 2,
        top["POT_AST_PG"],
        width,
        label="Potential assists",
        color="#2563eb"
    )

    ax.bar(
        x + width / 2,
        top["NBA_APG"],
        width,
        label="Actual assists",
        color="#f97316"
    )

    ax.set_xticks(x)
    ax.set_xticklabels(
        top["player"],
        rotation=25,
        ha="right"
    )

    ax.set_ylabel("Assists per game")
    ax.set_title(
        f"Minnesota creators: {phase}",
        fontweight="bold"
    )

    ax.legend()
    ax.grid(axis="y", alpha=0.15)

    fig.tight_layout()

    fig.savefig(
        FIG / f"03_minnesota_creators_{phase}.png",
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)


# ================================================================
# 11. BUILD LEAGUE-WIDE ESPN PLAYER STATISTICS
# ================================================================

league = (
    p.groupby(
        ["phase", "player"],
        as_index=False
    )
    .agg(
        GP=("game_id", "nunique"),
        MIN=("minutes", "sum"),
        PTS=("PTS", "sum"),
        AST=("AST", "sum"),
        TOV=("TOV", "sum")
    )
)

league["PPG"] = (
    league["PTS"] / league["GP"]
)

league["MPG"] = (
    league["MIN"] / league["GP"]
)

league["APG"] = (
    league["AST"] / league["GP"]
)

league["AST_TO"] = (
    league["AST"]
    / league["TOV"].replace(0, np.nan)
)

# This is a standardized 12-minute rate,
# NOT actual first-quarter assists.
league["AST_PER_12"] = (
    12 * league["AST"] / league["MIN"]
)

# Assign each player their most-played ESPN position.
# Players with no usable position are NOT silently
# assumed to be SGs.

if position_col is not None:

    positions = (
        p.dropna(subset=["position"])
        .groupby(
            ["phase", "player", "position"],
            as_index=False
        )["minutes"]
        .sum()
        .sort_values("minutes", ascending=False)
        .drop_duplicates(["phase", "player"])
    )

    league = league.merge(
        positions[
            ["phase", "player", "position"]
        ],
        on=["phase", "player"],
        how="left"
    )

else:
    league["position"] = np.nan

save_table(
    league,
    "league_espn_player_statistics.csv"
)





# ================================================================
# 15. MIKE CONLEY — MINNESOTA-ONLY STATISTICS
# ================================================================

conley_games = p[
    (p["team"] == "MIN")
    & (p["player"] == "Mike Conley")
].copy()

conley = (
    conley_games
    .groupby("phase", as_index=False)
    .agg(
        GP=("game_id", "nunique"),
        MIN=("minutes", "sum"),
        AST=("AST", "sum")
    )
)

conley["player"] = "Mike Conley"

conley["MPG"] = (
    conley["MIN"] / conley["GP"]
)

conley["APG"] = (
    conley["AST"] / conley["GP"]
)

conley["AST_PER_12"] = (
    12 * conley["AST"] / conley["MIN"]
)


conley_passing = min_passing[
    min_passing["player"] == "Mike Conley"
][
    ["phase", "POT_AST_PG"]
]

conley = conley.merge(
    conley_passing,
    on="phase",
    how="left"
)

print("\n6. MIKE CONLEY: MINNESOTA-ONLY")

save_table(
    conley[
        [
            "phase",
            "player",
            "GP",
            "MPG",
            "APG",
            "POT_AST_PG",
            "AST_PER_12"
        ]
    ],
    "06_conley_minnesota.csv"
)


# ================================================================
# 16. CONLEY VS NBA TOP FIVE ASSIST LEADERS
# ================================================================

# Leaders selected by actual APG, not potential assists.
#
# Regular season: minimum 58 appearances.
# Playoffs: minimum 5 appearances.
#
# League player totals combine teams when a player
# changed teams during the season.

for phase, minimum_games in [
    ("Regular Season", 58),
    ("Playoffs", 5)
]:

    leaders = (
        league[
            (league["phase"] == phase)
            & (league["GP"] >= minimum_games)
        ]
        .nlargest(5, "APG")
        .copy()
    )

    conley_phase = conley[
        conley["phase"] == phase
    ][
        [
            "player",
            "GP",
            "MPG",
            "APG",
            "AST_PER_12"
        ]
    ]

    comparison = (
        pd.concat(
            [leaders, conley_phase],
            ignore_index=True
        )
        .drop_duplicates(
            "player",
            keep="last"
        )
        .sort_values(
            "APG",
            ascending=False
        )
    )

    output = comparison[
        [
            "player",
            "GP",
            "MPG",
            "APG",
            "AST_PER_12"
        ]
    ]

    print(
        f"\n6. CONLEY VS NBA ASSIST LEADERS: {phase}"
    )

    save_table(
        output,
        f"06_conley_vs_leaders_{phase}.csv"
    )

    # Three separate charts.
    for metric, title in [
        ("APG", "Assists per game"),
        ("MPG", "Minutes per game"),
        ("AST_PER_12", "Assists per 12 minutes")
    ]:

        bar_chart(
            output,
            metric,
            f"{title}: Conley vs NBA leaders — {phase}",
            f"06_conley_{metric}_{phase}.png",
            highlight="Mike Conley"
        )


# ================================================================
# 17. FINAL VALIDATION AND OUTPUT SUMMARY
# ================================================================

print("\n" + "=" * 65)
print("CATEGORY 2 — FINAL VALIDATION")
print("=" * 65)

print("\nMinnesota games by phase:")
display(
    min_games.groupby("phase")["game_id"]
    .nunique()
    .reindex(PHASES)
)

print("\nEdwards appearances:")
display(
    ed.groupby("phase")["game_id"]
    .nunique()
    .reindex(PHASES)
)

print("\nConley Minnesota appearances:")
display(
    conley.set_index("phase")["GP"]
    .reindex(PHASES)
)

print("\nGenerated CSV files:")
for file in sorted(OUT.glob("*.csv")):
    print(" -", file.name)

print("\nGenerated charts:")
print(len(list(FIG.glob("*.png"))), "PNG files")



print("\nDONE.")
```

```python

# ============================================================
# EDWARDS VS TOP 5 SHOOTING GUARDS
# 2025–26 REGULAR SEASON AND PLAYOFFS
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. FILE PATHS

BASE = Path("/content/drive/MyDrive/minnesota_analysis")
RAW = BASE / "data/raw/player_box_2026.parquet"
OUT = BASE / "data/processed/category_2_final"
FIG = BASE / "figures/category_2_final"

OUT.mkdir(parents=True, exist_ok=True)
FIG.mkdir(parents=True, exist_ok=True)

# 2. LOAD AND STANDARDIZE RAW DATA

p = pd.read_parquet(RAW).copy()

p = p.rename(columns={
    "athlete_display_name": "player",
    "team_abbreviation": "team",
    "athlete_position_abbreviation": "position"
})

p["game_id"] = p["game_id"].astype(str)
p["season_type"] = pd.to_numeric(
    p["season_type"], errors="coerce"
)

# Remove the three identified All-Star games and NBA Cup final.
EXCLUDED_GAMES = {
    "401838140",
    "401838141",
    "401838143",
    "401809839"
}

p = p[
    ~p["game_id"].isin(EXCLUDED_GAMES)
    & p["season_type"].isin([2, 3])
].copy()

# Keep NBA teams only, excluding exhibition teams.
NBA_TEAMS = {
    "ATL", "BOS", "BKN", "CHA", "CHI", "CLE",
    "DAL", "DEN", "DET", "GS", "GSW", "HOU",
    "IND", "LAC", "LAL", "MEM", "MIA", "MIL",
    "MIN", "NO", "NOP", "NY", "NYK", "OKC",
    "ORL", "PHI", "PHX", "POR", "SAC", "SA",
    "SAS", "TOR", "UTAH", "UT", "UTA", "WSH",
    "WAS"
}

p = p[p["team"].isin(NBA_TEAMS)].copy()

# Remove players who did not participate.
p = p[
    p["did_not_play"].fillna(False).eq(False)
].copy()

# Convert box-score statistics to numbers.
STAT_COLUMNS = [
    "minutes",
    "points",
    "assists",
    "turnovers",
    "field_goals_made",
    "field_goals_attempted",
    "three_point_field_goals_made",
    "three_point_field_goals_attempted",
    "free_throws_made",
    "free_throws_attempted"
]

for col in STAT_COLUMNS:
    p[col] = pd.to_numeric(p[col], errors="coerce")

# Count only actual appearances.
p = p[p["minutes"] > 0].copy()

# Use athlete_id to avoid combining players with identical names.
p["athlete_id"] = p["athlete_id"].astype("Int64")

# Check for repeated records rather than silently deleting them.
duplicate_mask = p.duplicated(
    ["game_id", "athlete_id"], keep=False
)

if duplicate_mask.any():
    print("WARNING: Repeated player-game records found.")
    display(
        p.loc[
            duplicate_mask,
            ["game_id", "player", "team", "points"]
        ].sort_values(["player", "game_id"])
    )
    raise ValueError(
        "Resolve duplicate player-game records before aggregating."
    )

p["phase"] = p["season_type"].map({
    2: "Regular Season",
    3: "Playoffs"
})

# 3. VERIFY EDWARDS' REGULAR-SEASON DATA

ant = p[
    (p["player"] == "Anthony Edwards")
    & (p["team"] == "MIN")
    & (p["phase"] == "Regular Season")
]

print("EDWARDS REGULAR-SEASON AUDIT")
print("Games:", ant["game_id"].nunique())
print("Points:", ant["points"].sum())
print("PPG:", round(ant["points"].mean(), 2))
print("Assists:", ant["assists"].sum())
print("Turnovers:", ant["turnovers"].sum())

assert ant["game_id"].nunique() == 61, (
    "Edwards' game count is incorrect."
)

assert ant["points"].sum() == 1757, (
    "Edwards' point total is incorrect."
)

print("PASS: Edwards' games and points match the benchmark.")

# 4. BUILD PLAYER STATISTICS FROM GAME TOTALS

AGGREGATIONS = {
    "game_id": "nunique",
    "minutes": "sum",
    "points": "sum",
    "assists": "sum",
    "turnovers": "sum",
    "field_goals_made": "sum",
    "field_goals_attempted": "sum",
    "three_point_field_goals_made": "sum",
    "three_point_field_goals_attempted": "sum",
    "free_throws_made": "sum",
    "free_throws_attempted": "sum"
}

league = (
    p.groupby(
        ["phase", "athlete_id", "player"],
        as_index=False
    )
    .agg(AGGREGATIONS)
    .rename(columns={
        "game_id": "GP",
        "minutes": "MIN",
        "points": "PTS",
        "assists": "AST",
        "turnovers": "TOV",
        "field_goals_made": "FGM",
        "field_goals_attempted": "FGA",
        "three_point_field_goals_made": "FG3M",
        "three_point_field_goals_attempted": "FG3A",
        "free_throws_made": "FTM",
        "free_throws_attempted": "FTA"
    })
)

league["PPG"] = league["PTS"] / league["GP"]
league["MPG"] = league["MIN"] / league["GP"]
league["APG"] = league["AST"] / league["GP"]
league["TOV_PG"] = league["TOV"] / league["GP"]

league["AST_TO"] = (
    league["AST"] /
    league["TOV"].replace(0, np.nan)
)

league["FG_PCT"] = (
    100 * league["FGM"] /
    league["FGA"].replace(0, np.nan)
)

league["FG3_PCT"] = (
    100 * league["FG3M"] /
    league["FG3A"].replace(0, np.nan)
)

league["EFG_PCT"] = (
    100 * (league["FGM"] + 0.5 * league["FG3M"]) /
    league["FGA"].replace(0, np.nan)
)

league["TS_PCT"] = (
    100 * league["PTS"] /
    (2 * (league["FGA"] + 0.44 * league["FTA"]))
    .replace(0, np.nan)
)

# Save the complete corrected player table.
league.to_csv(
    OUT / "corrected_league_player_stats.csv",
    index=False
)

# 5. SHOOTING GUARD CANDIDATES

# This is the same broad SG / SG-hybrid candidate pool.
# Position classification is a methodological choice.
# Review hybrid PG/SG players before publishing.

SG_CANDIDATES = {
    "Anthony Edwards",
    "Donovan Mitchell",
    "Devin Booker",
    "Desmond Bane",
    "Derrick White",
    "Zach LaVine",
    "Jalen Green",
    "Tyler Herro",
    "Austin Reaves",
    "Norman Powell",
    "Bradley Beal",
    "CJ McCollum",
    "Anfernee Simons",
    "Cam Thomas",
    "Shaedon Sharpe",
    "Bennedict Mathurin",
    "Quentin Grimes",
    "VJ Edgecombe",
    "Kon Knueppel",
    "Nickeil Alexander-Walker",
    "Donte DiVincenzo",
    "Kentavious Caldwell-Pope",
    "Malik Monk",
    "Jordan Clarkson",
    "Grayson Allen",
    "Christian Braun",
    "De'Anthony Melton",
    "Keon Ellis",
    "Brandin Podziemski",
    "Ayo Dosunmu"
}

POSITION_REVIEW = {
    "Austin Reaves",
    "CJ McCollum",
    "Derrick White",
    "Brandin Podziemski",
    "Ayo Dosunmu"
}



passing = pd.DataFrame(
    [
        # Regular season
        ("Regular Season", "Anthony Edwards", 6.6),
        ("Regular Season", "Devin Booker", 11.9),
        ("Regular Season", "Donovan Mitchell", 10.6),
        ("Regular Season", "Derrick White", 9.5),
        ("Regular Season", "Desmond Bane", 7.3),
        ("Regular Season", "VJ Edgecombe", 7.4),
        ("Regular Season", "Kon Knueppel", 6.3),
        ("Regular Season", "CJ McCollum", 7.6),
        ("Regular Season", "Nickeil Alexander-Walker", 6.8),

        # Playoffs
        ("Playoffs", "Donovan Mitchell", 7.1),
        ("Playoffs", "Anthony Edwards", 5.7),
        ("Playoffs", "Austin Reaves", 11.5),
        ("Playoffs", "CJ McCollum", 4.2),
        ("Playoffs", "Desmond Bane", 4.4)
    ],
    columns=["phase", "player", "POT_AST_PG"]
)

# 7. CREATE BOTH COMPARISONS

PHASES = {
    "Regular Season": 58,
    "Playoffs": 4
}

CHARTS = {
    "PPG": "Points per game",
    "POT_AST_PG": "Potential assists per game",
    "APG": "Assists per game",
    "AST_TO": "Assist-to-turnover ratio"
}

results = {}

for phase, min_games in PHASES.items():

    eligible = league[
        (league["phase"] == phase)
        & league["player"].isin(SG_CANDIDATES)
        & (league["GP"] >= min_games)
    ].copy()

    # Rank using corrected ESPN points per game.
    eligible = eligible.sort_values(
        ["PPG", "player"],
        ascending=[False, True]
    )

    top_five = eligible.head(5).copy()

    # Include Edwards even if outside the top five.
    edwards = league[
        (league["phase"] == phase)
        & (league["player"] == "Anthony Edwards")
    ].copy()

    comparison = (
        pd.concat([top_five, edwards])
        .drop_duplicates(["phase", "athlete_id"])
        .copy()
    )

    comparison["TOP_FIVE"] = (
        comparison["athlete_id"].isin(
            top_five["athlete_id"]
        )
    )

    comparison["POSITION_REVIEW"] = (
        comparison["player"].isin(POSITION_REVIEW)
    )

    comparison = comparison.merge(
        passing[passing["phase"] == phase][
            ["player", "POT_AST_PG"]
        ],
        on="player",
        how="left",
        validate="many_to_one"
    )

    comparison = comparison.sort_values(
        "PPG", ascending=False
    ).reset_index(drop=True)

    output_columns = [
        "player",
        "GP",
        "MPG",
        "PPG",
        "APG",
        "TOV_PG",
        "AST_TO",
        "POT_AST_PG",
        "FG_PCT",
        "FG3_PCT",
        "EFG_PCT",
        "TS_PCT",
        "TOP_FIVE",
        "POSITION_REVIEW"
    ]

    result = comparison[output_columns].copy()

    results[phase] = result

    print("\n" + "=" * 65)
    print(f"{phase.upper()} — EDWARDS VS TOP FIVE SGs")
    print(f"Minimum games: {min_games}")
    print("=" * 65)

    display(result.round(2))

    safe_phase = phase.lower().replace(" ", "_")

    result.to_csv(
        OUT / f"edwards_top5_sg_{safe_phase}.csv",
        index=False
    )

    # 8. CREATE CHARTS FOR EACH PHASE

    for metric, label in CHARTS.items():

        chart_data = (
            result[["player", metric]]
            .dropna(subset=[metric])
            .sort_values(metric)
        )

        if chart_data.empty:
            continue

        fig, ax = plt.subplots(figsize=(10, 5))

        colors = [
            "#f97316"
            if name == "Anthony Edwards"
            else "#2563eb"
            for name in chart_data["player"]
        ]

        bars = ax.barh(
            chart_data["player"],
            chart_data[metric],
            color=colors
        )

        max_value = chart_data[metric].max()
        ax.set_xlim(0, max(1, max_value * 1.25))

        for bar, value in zip(
            bars, chart_data[metric]
        ):
            ax.text(
                bar.get_width() + max_value * 0.02,
                bar.get_y() + bar.get_height() / 2,
                f"{value:.2f}",
                va="center"
            )

        ax.set_title(
            f"Edwards vs Top 5 SGs — {label}\n{phase}"
        )
        ax.set_xlabel(label)
        ax.grid(axis="x", alpha=0.2)
        ax.set_axisbelow(True)

        plt.tight_layout()

        fig.savefig(
            FIG / f"sg_{safe_phase}_{metric.lower()}.png",
            dpi=200,
            bbox_inches="tight"
        )

        plt.show()
        plt.close(fig)


    missing_potential = result.loc[
        result["POT_AST_PG"].isna(), "player"
    ].tolist()

    if missing_potential:
        print(
            "\nPotential-assist values still needed:",
            ", ".join(missing_potential)
        )

    review = result.loc[
        result["POSITION_REVIEW"], "player"
    ].tolist()

    if review:
        print(
            "Hybrid positions to review:",
            ", ".join(review)
        )

# 9. FINAL CHECK

corrected_ant = results["Regular Season"].query(
    "player == 'Anthony Edwards'"
).iloc[0]

assert corrected_ant["GP"] == 61
assert round(corrected_ant["PPG"], 1) == 28.8

print("\nSUCCESS")
print(
    "Edwards:",
    corrected_ant["GP"],
    "regular-season games,",
    round(corrected_ant["PPG"], 2),
    "PPG"
)
print("Comparison tables:", OUT)
print("Charts:", FIG)
```
