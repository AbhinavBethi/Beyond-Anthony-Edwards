```python
from google.colab import drive
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests

drive.mount("/content/drive")

ROOT = Path("/content/drive/MyDrive/minnesota_analysis")
RAW = ROOT / "data" / "raw"
OUT = ROOT / "data" / "processed" / "category_1_final"
FIG = ROOT / "figures" / "category_1_final"

for folder in [RAW, OUT, FIG]:
    folder.mkdir(parents=True, exist_ok=True)

sns.set_theme(style="whitegrid")

SEASONS = [2024, 2025, 2026]
FINAL_FOUR = ["NY", "CLE", "OKC", "SA"]
COMPARISON_TEAMS = ["MIN"] + FINAL_FOUR

# Confirmed extra NBA Cup championship game IDs.
# They do not count toward regular-season standings.
CUP_FINAL_IDS = {
    "2024-25": "401734908",
    "2025-26": "401809839",
}

NBA_TEAMS = {
    "ATL", "BOS", "BKN", "CHA", "CHI",
    "CLE", "DAL", "DEN", "DET", "GS",
    "HOU", "IND", "LAC", "LAL", "MEM",
    "MIA", "MIL", "MIN", "NO", "NY",
    "OKC", "ORL", "PHI", "PHX", "POR",
    "SAC", "SA", "TOR", "UTAH", "WSH"
}

COLS = {
    "points": "team_score",
    "fgm": "field_goals_made",
    "fga": "field_goals_attempted",
    "three_pm": "three_point_field_goals_made",
    "fta": "free_throws_attempted",
    "orb": "offensive_rebounds",
    "drb": "defensive_rebounds",
    "tov": "turnovers",
}

METRICS = {
    "ortg": "Offensive Rating",
    "drtg": "Defensive Rating",
    "net_rating": "Net Rating",
    "efg_pct": "Effective FG%",
    "tov_pct": "Turnover Percentage",
    "orb_pct": "Offensive Rebounding Percentage",
    "pace_proxy": "Estimated Possessions per Game",
}

PERCENT = {"efg_pct", "tov_pct", "orb_pct"}


# ============================================================
# 1. LOAD ALL THREE SEASONS
# ============================================================

def load_season(year):
    path = RAW / f"team_box_{year}.parquet"

    if not path.exists():
        url = (
            "https://github.com/sportsdataverse/"
            "sportsdataverse-data/releases/download/"
            "espn_nba_team_boxscores/"
            f"team_box_{year}.parquet"
        )

        print("Downloading:", year)

        response = requests.get(url, timeout=120)
        response.raise_for_status()
        path.write_bytes(response.content)

    frame = pd.read_parquet(path)
    frame["season"] = f"{year-1}-{str(year)[-2:]}"
    return frame


raw = pd.concat(
    [load_season(year) for year in SEASONS],
    ignore_index=True
)

print("Loaded rows:", len(raw))


# ============================================================
# 2. STANDARDIZE AND CLEAN
# ============================================================

required = [
    "game_id", "game_date", "season_type",
    "team_abbreviation"
] + list(COLS.values())

missing = [c for c in required if c not in raw.columns]

if missing:
    raise ValueError(f"Missing source columns: {missing}")

df = raw.copy()

df["game_id"] = df["game_id"].astype(str)

df["team"] = (
    df["team_abbreviation"]
    .astype(str)
    .str.upper()
    .str.strip()
)

df["date"] = pd.to_datetime(
    df["game_date"],
    errors="coerce",
    utc=True
).dt.tz_localize(None)

df["season_type"] = pd.to_numeric(
    df["season_type"],
    errors="coerce"
)

# ESPN season type 2 = regular season.
# ESPN season type 3 = playoffs.
df = df[
    df["season_type"].isin([2, 3])
    & df["team"].isin(NBA_TEAMS)
].copy()

df["phase"] = np.where(
    df["season_type"] == 2,
    "Regular Season",
    "Playoffs"
)

for metric, source in COLS.items():
    df[metric] = pd.to_numeric(
        df[source],
        errors="coerce"
    )

if df[list(COLS)].isna().any().any():
    display(
        df.groupby(["season", "phase"])[list(COLS)]
        .apply(lambda x: x.isna().sum())
    )
    raise ValueError("Missing box-score statistics.")

if df["date"].isna().any():
    raise ValueError("Unparseable game dates.")

if df.duplicated(
    ["season", "phase", "game_id", "team"]
).any():
    raise ValueError("Duplicate team-game rows.")


# ============================================================
# 3. REMOVE NBA CUP FINALS
# ============================================================

regular = df[df["phase"] == "Regular Season"].copy()

# Remove confirmed 2024-25 and 2025-26 Cup finals.
for season, game_id in CUP_FINAL_IDS.items():

    cup = regular[
        (regular["season"] == season)
        & (regular["game_id"] == game_id)
    ]

    print(
        f"{season} Cup-final rows:",
        len(cup)
    )

    if len(cup) != 2:
        raise ValueError(
            f"Could not identify Cup final {game_id}."
        )

    regular = regular[
        ~(
            (regular["season"] == season)
            & (regular["game_id"] == game_id)
        )
    ].copy()

# The 2023-24 Cup final was LAL vs IND,
# played December 9, 2023.
cup_2024 = regular[
    (regular["season"] == "2023-24")
    & (regular["date"].dt.date == pd.Timestamp(
        "2023-12-09"
    ).date())
    & regular["team"].isin(["LAL", "IND"])
]

cup_2024_games = (
    cup_2024.groupby("game_id")["team"]
    .agg(lambda x: set(x))
)

cup_2024_ids = cup_2024_games[
    cup_2024_games.apply(
        lambda x: x == {"LAL", "IND"}
    )
].index.tolist()

if len(cup_2024_ids) != 1:
    raise ValueError(
        "Could not uniquely identify the "
        "2023-24 NBA Cup championship."
    )

regular = regular[
    ~(
        (regular["season"] == "2023-24")
        & (regular["game_id"] == cup_2024_ids[0])
    )
].copy()

# Validate every regular-season team has 82 games.
coverage = regular.groupby(
    ["season", "team"]
)["game_id"].nunique()

print("\nREGULAR-SEASON COVERAGE")
display(coverage.unstack("season"))

if (
    len(coverage) != 90
    or not (coverage == 82).all()
):
    raise ValueError(
        "Regular-season coverage is incomplete."
    )

print("All three regular seasons validated.")


# ============================================================
# 4. PREPARE PLAYOFF DATA
# ============================================================

playoffs = df[df["phase"] == "Playoffs"].copy()

print("\nPLAYOFF GAME COUNTS")
playoff_coverage = playoffs.groupby(
    ["season", "team"]
)["game_id"].nunique()

display(
    playoff_coverage.unstack("season")
)

# Confirm Minnesota has playoff data in all 3 years.
for season in ["2023-24", "2024-25", "2025-26"]:

    n = len(
        playoffs[
            (playoffs["season"] == season)
            & (playoffs["team"] == "MIN")
        ]
    )

    if n == 0:
        raise ValueError(
            f"No Minnesota playoff data for {season}."
        )

# Check that all four 2026 finalists have
# playoff box scores.
for team in FINAL_FOUR:

    n = len(
        playoffs[
            (playoffs["season"] == "2025-26")
            & (playoffs["team"] == team)
        ]
    )

    if n == 0:
        raise ValueError(
            f"No 2026 playoff data for {team}."
        )


# ============================================================
# 5. MATCH TEAMS TO OPPONENTS
# ============================================================

all_games = pd.concat(
    [regular, playoffs],
    ignore_index=True
)

STAT_COLS = list(COLS)

team_games = all_games[
    [
        "season", "phase", "game_id",
        "date", "team"
    ] + STAT_COLS
].copy()

opponent = team_games[
    [
        "season", "phase", "game_id",
        "team"
    ] + STAT_COLS
].rename(
    columns={
        "team": "opponent",
        **{
            c: "opp_" + c
            for c in STAT_COLS
        }
    }
)

paired = team_games.merge(
    opponent,
    on=["season", "phase", "game_id"],
    how="inner"
)

paired = paired[
    paired["team"] != paired["opponent"]
].copy()

if paired.duplicated(
    ["season", "phase", "game_id", "team"]
).any():
    raise ValueError("Opponent join produced duplicates.")

# Every retained game must have two teams.
expected_rows = len(team_games)

if len(paired) != expected_rows:
    raise ValueError(
        "Some games are missing opponent records."
    )


# ============================================================
# 6. GAME-LEVEL STATISTICS
# ============================================================

paired["team_poss"] = (
    paired["fga"]
    - paired["orb"]
    + paired["tov"]
    + 0.44 * paired["fta"]
)

paired["opp_poss"] = (
    paired["opp_fga"]
    - paired["opp_orb"]
    + paired["opp_tov"]
    + 0.44 * paired["opp_fta"]
)

paired["possessions"] = (
    paired["team_poss"]
    + paired["opp_poss"]
) / 2

if (paired["possessions"] <= 0).any():
    raise ValueError("Invalid possession estimates.")

paired["win"] = (
    paired["points"] > paired["opp_points"]
).astype(int)

paired["loss"] = (
    paired["points"] < paired["opp_points"]
).astype(int)

paired["game_ortg"] = (
    100 * paired["points"]
    / paired["possessions"]
)

paired["game_drtg"] = (
    100 * paired["opp_points"]
    / paired["possessions"]
)

paired["game_net_rating"] = (
    paired["game_ortg"]
    - paired["game_drtg"]
)


# ============================================================
# 7. AGGREGATION FUNCTION
# ============================================================

def summarize(frame, group_cols):

    result = frame.groupby(
        group_cols,
        as_index=False
    ).agg(
        games=("game_id", "nunique"),
        wins=("win", "sum"),
        losses=("loss", "sum"),

        points=("points", "sum"),
        opp_points=("opp_points", "sum"),

        possessions=("possessions", "sum"),

        fgm=("fgm", "sum"),
        fga=("fga", "sum"),
        three_pm=("three_pm", "sum"),
        fta=("fta", "sum"),

        orb=("orb", "sum"),
        tov=("tov", "sum"),
        opp_drb=("opp_drb", "sum")
    )

    result["win_pct"] = (
        result["wins"] / result["games"]
    )

    result["ortg"] = (
        100 * result["points"]
        / result["possessions"]
    )

    result["drtg"] = (
        100 * result["opp_points"]
        / result["possessions"]
    )

    result["net_rating"] = (
        result["ortg"] - result["drtg"]
    )

    result["efg_pct"] = (
        result["fgm"]
        + 0.5 * result["three_pm"]
    ) / result["fga"]

    result["tov_pct"] = (
        result["tov"]
        / (
            result["fga"]
            + 0.44 * result["fta"]
            + result["tov"]
        )
    )

    result["orb_pct"] = (
        result["orb"]
        / (
            result["orb"]
            + result["opp_drb"]
        )
    )

    # Estimated possessions per game.
    # This is a pace proxy, not official
    # possessions per 48 minutes.
    result["pace_proxy"] = (
        result["possessions"]
        / result["games"]
    )

    result["point_diff"] = (
        result["points"]
        - result["opp_points"]
    ) / result["games"]

    return result


team_stats = summarize(
    paired,
    ["season", "phase", "team"]
)

team_stats.to_csv(
    OUT / "all_team_regular_and_playoff_stats.csv",
    index=False
)


# ============================================================
# PART A — MINNESOTA'S THREE PLAYOFF RUNS
# ============================================================

min_playoffs = team_stats[
    (team_stats["team"] == "MIN")
    & (team_stats["phase"] == "Playoffs")
].sort_values("season").copy()

print("\nMINNESOTA: THREE PLAYOFF RUNS")

display(
    min_playoffs[
        [
            "season", "games", "wins", "losses",
            "ortg", "drtg", "net_rating",
            "efg_pct", "tov_pct", "orb_pct",
            "pace_proxy"
        ]
    ].round(3)
)

min_playoffs.to_csv(
    OUT / "minnesota_three_playoff_runs.csv",
    index=False
)

# Playoff performance charts for all 3 years.
for metric, label in METRICS.items():

    chart = min_playoffs.copy()

    if metric in PERCENT:
        chart[metric] *= 100

    fig, ax = plt.subplots(figsize=(9, 5))

    bars = ax.bar(
        chart["season"],
        chart[metric],
        color="#2563eb"
    )

    ax.bar_label(
        bars,
        fmt="%.1f",
        padding=3
    )

    ax.set_title(
        f"Minnesota Playoffs: 2024–2026\n{label}"
    )

    ax.set_ylabel(
        "Percentage (%)"
        if metric in PERCENT
        else label
    )

    if metric == "net_rating":
        ax.axhline(0, color="black")

    fig.tight_layout()

    fig.savefig(
        FIG / f"min_playoffs_{metric}.png",
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)


# ============================================================
# PART B — REGULAR SEASON VS PLAYOFFS
# ============================================================

min_all = team_stats[
    team_stats["team"] == "MIN"
].copy()

min_all.to_csv(
    OUT / "minnesota_regular_vs_playoffs.csv",
    index=False
)

print("\nMINNESOTA: REGULAR SEASON VS PLAYOFFS")

display(
    min_all[
        [
            "season", "phase", "games",
            "ortg", "drtg", "net_rating",
            "efg_pct", "tov_pct", "orb_pct"
        ]
    ].round(3)
)

# Calculate playoff minus regular-season gaps.
min_wide = min_all.pivot(
    index="season",
    columns="phase",
    values=list(METRICS)
)

gap_rows = []

for season in sorted(min_all["season"].unique()):

    row = {"season": season}

    for metric in METRICS:

        reg = min_wide.loc[
            season,
            (metric, "Regular Season")
        ]

        post = min_wide.loc[
            season,
            (metric, "Playoffs")
        ]

        multiplier = (
            100 if metric in PERCENT else 1
        )

        row[metric] = (
            post - reg
        ) * multiplier

    gap_rows.append(row)

min_gaps = pd.DataFrame(gap_rows)

print("\nPLAYOFF MINUS REGULAR-SEASON GAPS")

display(min_gaps.round(2))

min_gaps.to_csv(
    OUT / "minnesota_playoff_change_by_season.csv",
    index=False
)

# Compare both phases across three seasons.
for metric, label in METRICS.items():

    chart = min_all.copy()

    if metric in PERCENT:
        chart[metric] *= 100

    fig, ax = plt.subplots(figsize=(10, 5))

    sns.barplot(
        data=chart,
        x="season",
        y=metric,
        hue="phase",
        palette={
            "Regular Season": "#94a3b8",
            "Playoffs": "#2563eb"
        },
        ax=ax
    )

    ax.set_title(
        f"Minnesota: Regular Season vs Playoffs\n"
        f"{label}"
    )

    ax.set_xlabel("Season")

    ax.set_ylabel(
        "Percentage (%)"
        if metric in PERCENT
        else label
    )

    ax.legend(title="")

    fig.tight_layout()

    fig.savefig(
        FIG / f"min_reg_vs_playoffs_{metric}.png",
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)


# ============================================================
# PART C — 2026 CONFERENCE FINALISTS
# ============================================================

# Four teams that reached the 2026
# Eastern and Western Conference Finals:
#
# NY, CLE, OKC, SA
#
# Compare them with Minnesota in BOTH phases.

finalists = team_stats[
    (team_stats["season"] == "2025-26")
    & team_stats["team"].isin(COMPARISON_TEAMS)
].copy()

expected = {
    (team, phase)
    for team in COMPARISON_TEAMS
    for phase in ["Regular Season", "Playoffs"]
}

actual = set(
    zip(finalists["team"], finalists["phase"])
)

if actual != expected:
    raise ValueError(
        f"Missing team-phase combinations: "
        f"{expected - actual}"
    )

finalists.to_csv(
    OUT / "2026_minnesota_vs_final_four.csv",
    index=False
)

print("\n2026: MINNESOTA VS CONFERENCE FINALISTS")

display(
    finalists[
        [
            "team", "phase", "games",
            "wins", "losses",
            "ortg", "drtg", "net_rating",
            "efg_pct", "tov_pct", "orb_pct"
        ]
    ].sort_values(
        ["phase", "team"]
    ).round(3)
)


# ============================================================
# 8. FINALISTS — REGULAR SEASON VS PLAYOFFS
# ============================================================

for metric, label in METRICS.items():

    chart = finalists.copy()

    if metric in PERCENT:
        chart[metric] *= 100

    fig, ax = plt.subplots(
        figsize=(11, 5)
    )

    sns.barplot(
        data=chart,
        x="team",
        y=metric,
        hue="phase",
        order=COMPARISON_TEAMS,
        hue_order=[
            "Regular Season",
            "Playoffs"
        ],
        palette={
            "Regular Season": "#94a3b8",
            "Playoffs": "#2563eb"
        },
        ax=ax
    )

    ax.set_title(
        "2025–26: Minnesota vs "
        "Four Conference Finalists\n"
        + label
    )

    ax.set_ylabel(
        "Percentage (%)"
        if metric in PERCENT
        else label
    )

    ax.set_xlabel("Team")
    ax.legend(title="")

    fig.tight_layout()

    fig.savefig(
        FIG / f"2026_finalists_{metric}.png",
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)


# ============================================================
# 9. FINAL FOUR AVERAGE VS MINNESOTA
# ============================================================

# Each finalist receives equal weight.
# Regular season and playoffs are calculated
# separately.

four_avg = (
    finalists[
        finalists["team"].isin(FINAL_FOUR)
    ]
    .groupby("phase")[list(METRICS)]
    .mean()
)

min_2026 = (
    finalists[
        finalists["team"] == "MIN"
    ]
    .set_index("phase")
)

final_gap_rows = []

for phase in [
    "Regular Season",
    "Playoffs"
]:

    for metric, label in METRICS.items():

        min_value = min_2026.loc[
            phase, metric
        ]

        avg_value = four_avg.loc[
            phase, metric
        ]

        multiplier = (
            100 if metric in PERCENT else 1
        )

        final_gap_rows.append({
            "phase": phase,
            "metric": label,
            "minnesota": min_value * multiplier,
            "final_four_average": (
                avg_value * multiplier
            ),
            "gap": (
                min_value - avg_value
            ) * multiplier
        })

final_gaps = pd.DataFrame(
    final_gap_rows
)

print(
    "\n2026: MINNESOTA MINUS "
    "FOUR FINALISTS' AVERAGE"
)

display(
    final_gaps.round(2)
)

final_gaps.to_csv(
    OUT / "2026_final_four_average_gaps.csv",
    index=False
)


# ============================================================
# 10. FINAL GAP HEATMAP
# ============================================================

heatmap = final_gaps.pivot(
    index="metric",
    columns="phase",
    values="gap"
)

heatmap = heatmap[
    ["Regular Season", "Playoffs"]
]

fig, ax = plt.subplots(
    figsize=(9, 7)
)

sns.heatmap(
    heatmap,
    annot=True,
    fmt="+.2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    ax=ax
)

ax.set_title(
    "2025–26: Minnesota vs "
    "Four Conference Finalists\n"
    "Minnesota Minus Four-Team Average"
)

ax.set_xlabel("")
ax.set_ylabel("")

fig.tight_layout()

fig.savefig(
    FIG / "2026_final_four_gap_heatmap.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# ============================================================
# 11. MINNESOTA PLAYOFF GAME LOGS
# ============================================================

min_games = paired[
    (paired["team"] == "MIN")
    & (paired["phase"] == "Playoffs")
].copy()

min_games = min_games.sort_values(
    ["season", "date"]
)

min_games["playoff_game_number"] = (
    min_games.groupby("season")
    .cumcount() + 1
)

min_games.to_csv(
    OUT / "minnesota_playoff_game_logs.csv",
    index=False
)

# A compact series-level summary for context,
# without detailed opponent-specific analysis.

series_summary = summarize(
    min_games,
    ["season", "opponent"]
)

print("\nMINNESOTA PLAYOFF SERIES SUMMARY")

display(
    series_summary[
        [
            "season", "opponent",
            "games", "wins", "losses",
            "ortg", "drtg", "net_rating"
        ]
    ].sort_values(
        ["season", "opponent"]
    ).round(2)
)

series_summary.to_csv(
    OUT / "minnesota_playoff_series_summary.csv",
    index=False
)


# ============================================================
# 12. COMPLETION SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("CATEGORY 1 ANALYSIS COMPLETE")
print("=" * 60)

print("\nCompleted:")
print("1. Minnesota's three playoff runs")
print("2. Regular season vs playoffs")
print("3. Playoff efficiency changes")
print("4. 2026 Minnesota vs four conference finalists")
print("5. Final-four average comparison")
print("6. Minnesota playoff series summary")

print("\nDatasets saved to:")
print(OUT)

print("\nCharts saved to:")
print(FIG)

print(
    "\nMethodology: ratings use estimated "
    "possessions, not official NBA ratings."
)

print(
    "The possession-per-game metric is a "
    "pace proxy, not official pace per 48 minutes."
)
```
